# Cheatsheet — Vectores
**Fundamentos matemáticos para Física · 1.º de Bachillerato**

## 1. Magnitudes escalares y vectoriales

**Magnitud escalar:** queda definida por un número y una unidad. Ejemplos: masa, tiempo, temperatura y energía.

**Magnitud vectorial:** necesita módulo, dirección y sentido. Ejemplos: desplazamiento, velocidad, aceleración y fuerza.

Un vector se representa mediante una flecha:

$$
\vec a
$$

Sus características son:

- **Módulo:** longitud o valor del vector.
- **Dirección:** recta sobre la que actúa.
- **Sentido:** hacia dónde apunta.
- **Punto de aplicación:** dónde se aplica, cuando es relevante.

---

## 2. Componentes de un vector

### En dos dimensiones (2D)

$$
\boxed{\vec a=a_x\vec i+a_y\vec j}
$$

También se escribe:

$$
\vec a=(a_x,a_y)
$$

Ejemplo:

$$
\vec a=3\vec i+5\vec j=(3,5)
$$

### En tres dimensiones (3D)

Añadimos la componente del eje $z$:

$$
\boxed{\vec a=a_x\vec i+a_y\vec j+a_z\vec k}
$$

Ejemplo:

$$
\vec a=3\vec i+5\vec j+\vec k
$$

$$
\boxed{\vec a=(3,5,1)}
$$

Los vectores $\vec i$, $\vec j$ y $\vec k$ son los vectores unitarios de los tres ejes:

$$
\vec i=(1,0,0)
$$

$$
\vec j=(0,1,0)
$$

$$
\vec k=(0,0,1)
$$

**Importante:** si no aparece un coeficiente delante de $\vec k$, su componente vale $1$. Si no aparece un eje, su componente vale $0$.

---

## 3. Módulo de un vector

El módulo es la longitud del vector.

### En 2D

$$
\boxed{|\vec a|=\sqrt{a_x^2+a_y^2}}
$$

Ejemplo:

$$
\vec a=(3,4)
$$

$$
|\vec a|=\sqrt{3^2+4^2}=\boxed{5}
$$

### En 3D

$$
\boxed{|\vec a|=\sqrt{a_x^2+a_y^2+a_z^2}}
$$

Ejemplo:

$$
\vec a=2\vec i+3\vec j+6\vec k
$$

$$
|\vec a|=\sqrt{2^2+3^2+6^2}=\boxed{7}
$$

---

## 4. Vector entre dos puntos

Si conocemos dos puntos $A$ y $B$, el vector que va desde $A$ hasta $B$ se calcula restando sus coordenadas:

$$
\boxed{\overrightarrow{AB}=B-A}
$$

En 2D:

$$
\overrightarrow{AB}=(x_B-x_A,\ y_B-y_A)
$$

En 3D:

$$
\overrightarrow{AB}=(x_B-x_A,\ y_B-y_A,\ z_B-z_A)
$$

**Ejemplo**

$$
A=(1,2,3),\qquad B=(4,6,5)
$$

$$
\overrightarrow{AB}=(4-1,6-2,5-3)
$$

$$
\boxed{\overrightarrow{AB}=(3,4,2)}
$$

El vector contrario es:

$$
\overrightarrow{BA}=(-3,-4,-2)
$$

---

## 5. Vector unitario y vector opuesto

### Vector unitario

Tiene módulo $1$ y permite representar una dirección.

Para obtener el vector unitario en la dirección de $\vec a$:

$$
\boxed{\hat a=\frac{\vec a}{|\vec a|}}
$$

Ejemplo:

$$
\vec a=(3,4),\qquad |\vec a|=5
$$

$$
\boxed{\hat a=\left(\frac35,\frac45\right)}
$$

### Vector opuesto

Tiene el mismo módulo y dirección, pero sentido contrario.

$$
\boxed{-\vec a=(-a_x,-a_y,-a_z)}
$$

Además:

$$
\vec a+(-\vec a)=\vec 0
$$

---

## 6. Suma y resta de vectores

Se opera componente a componente.

### Suma

$$
\boxed{\vec a+\vec b=(a_x+b_x,a_y+b_y,a_z+b_z)}
$$

### Resta

$$
\boxed{\vec a-\vec b=(a_x-b_x,a_y-b_y,a_z-b_z)}
$$

**Ejemplo**

$$
\vec a=(3,5,1),\qquad\vec b=(2,-1,4)
$$

$$
\vec a+\vec b=(5,4,5)
$$

$$
\vec a-\vec b=(1,6,-3)
$$

En Física, la suma de fuerzas permite calcular la fuerza resultante:

$$
\boxed{\vec F_R=\vec F_1+\vec F_2+\cdots}
$$

---

## 7. Multiplicación por un escalar

Multiplicar un vector por un número $k$ consiste en multiplicar todas sus componentes.

$$
\boxed{k\vec a=(ka_x,ka_y,ka_z)}
$$

Ejemplo:

$$
\vec a=(2,-1,3)
$$

$$
3\vec a=(6,-3,9)
$$

El módulo cambia según:

$$
\boxed{|k\vec a|=|k||\vec a|}
$$

- Si $k>0$, conserva el sentido.
- Si $k<0$, invierte el sentido.
- Si $|k|>1$, aumenta el módulo.
- Si $0<|k|<1$, disminuye el módulo.

---

## 8. De módulo y ángulo a componentes

Si un vector $\vec a$ tiene módulo $a$ y forma un ángulo $\alpha$ con el eje horizontal positivo:

$$
\boxed{a_x=a\cos\alpha}
$$

$$
\boxed{a_y=a\sin\alpha}
$$

Por tanto:

$$
\boxed{\vec a=(a\cos\alpha,a\sin\alpha)}
$$

**Ejemplo**

Una fuerza de $10\,N$ forma $30^\circ$ con la horizontal.

$$
F_x=10\cos30^\circ=8.66\,N
$$

$$
F_y=10\sin30^\circ=5\,N
$$

$$
\boxed{\vec F=(8.66,5)\,N}
$$

**Atención:** si el ángulo se mide respecto al eje vertical, el coseno corresponde a la componente vertical y el seno a la horizontal. Los signos dependen del sentido del vector.

---

## 9. De componentes a módulo y ángulo

Si conocemos:

$$
\vec a=(a_x,a_y)
$$

### Calcular el módulo

$$
\boxed{a=\sqrt{a_x^2+a_y^2}}
$$

### Calcular el ángulo

$$
\tan\alpha=\frac{a_y}{a_x}
$$

$$
\boxed{\alpha=\arctan\left(\frac{a_y}{a_x}\right)}
$$

Hay que comprobar el cuadrante para obtener la dirección correcta. Si la calculadora dispone de la función $\operatorname{atan2}(a_y,a_x)$, esta tiene en cuenta los signos de ambas componentes.

**Ejemplo**

$$
\vec a=(3,4)
$$

$$
a=\sqrt{3^2+4^2}=5
$$

$$
\alpha=\arctan\left(\frac43\right)\approx53.13^\circ
$$

Por tanto, el vector tiene módulo $5$ y forma aproximadamente $53.13^\circ$ con el eje $x$ positivo.

---

## 10. Calcular un vector conocido otro y el ángulo entre ambos

Supongamos que conocemos:

- Un vector $\vec a=(a_x,a_y)$.
- El módulo $b$ del segundo vector.
- El ángulo $\theta$ que forman ambos.

Queremos calcular $\vec b$.

### Paso 1. Calcular la dirección del primer vector

$$
\boxed{\alpha=\operatorname{atan2}(a_y,a_x)}
$$

### Paso 2. Calcular la dirección del segundo vector

Si giramos en sentido antihorario:

$$
\boxed{\beta=\alpha+\theta}
$$

Si giramos en sentido horario:

$$
\boxed{\beta=\alpha-\theta}
$$

### Paso 3. Calcular las componentes

$$
\boxed{b_x=b\cos\beta}
$$

$$
\boxed{b_y=b\sin\beta}
$$

### Ejemplo completo

Conocemos:

$$
\vec a=(3,4)
$$

Buscamos un vector de módulo $10$ que forme $30^\circ$ con $\vec a$, girando en sentido antihorario.

**Primero**, calculamos la dirección de $\vec a$:

$$
\alpha=\arctan\left(\frac43\right)=53.13^\circ
$$

**Segundo**, sumamos el ángulo:

$$
\beta=53.13^\circ+30^\circ=83.13^\circ
$$

**Tercero**, calculamos las componentes:

$$
b_x=10\cos83.13^\circ\approx1.20
$$

$$
b_y=10\sin83.13^\circ\approx9.93
$$

**Resultado:**

$$
\boxed{\vec b\approx(1.20,9.93)}
$$

### ¿Y si no indican el sentido del giro?

En 2D existen dos soluciones posibles:

$$
\boxed{\beta=\alpha\pm\theta}
$$

En este ejemplo, el giro horario daría aproximadamente:

$$
\boxed{\vec b\approx(9.20,3.93)}
$$

**Importante:** conocer únicamente un vector y el ángulo entre ambos no basta. También necesitamos el módulo del segundo vector u otra condición que permita determinarlo.

En 3D, conocer el módulo y el ángulo tampoco basta para determinar una dirección única: se necesita información adicional, como el plano de giro.

## 11. Producto escalar

El producto escalar de dos vectores da como resultado **un número**.

### Por componentes

En 2D:

$$
\boxed{\vec a\cdot\vec b=a_xb_x+a_yb_y}
$$

En 3D:

$$
\boxed{\vec a\cdot\vec b=a_xb_x+a_yb_y+a_zb_z}
$$

### Mediante el ángulo

$$
\boxed{\vec a\cdot\vec b=|\vec a||\vec b|\cos\theta}
$$

**Ejemplo**

$$
\vec a=(3,5,1),\qquad\vec b=(2,-1,4)
$$

$$
\vec a\cdot\vec b=3(2)+5(-1)+1(4)
$$

$$
\boxed{\vec a\cdot\vec b=5}
$$

### Calcular el ángulo entre dos vectores

Despejamos el coseno:

$$
\boxed{\cos\theta=\frac{\vec a\cdot\vec b}{|\vec a||\vec b|}}
$$

Y aplicamos arccoseno:

$$
\boxed{\theta=\arccos\left(\frac{\vec a\cdot\vec b}{|\vec a||\vec b|}\right)}
$$

Esta fórmula funciona tanto en 2D como en 3D, siempre que ninguno de los vectores sea nulo.

---

## 12. Vectores paralelos y perpendiculares

### Paralelos

Dos vectores no nulos son paralelos si uno es múltiplo del otro:

$$
\boxed{\vec a=k\vec b}
$$

Ejemplo:

$$
\vec a=(2,4,6),\qquad\vec b=(1,2,3)
$$

Como $\vec a=2\vec b$, son paralelos.

### Perpendiculares

Dos vectores no nulos son perpendiculares cuando su producto escalar es cero:

$$
\boxed{\vec a\cdot\vec b=0}
$$

Ejemplo:

$$
\vec a=(2,1),\qquad\vec b=(1,-2)
$$

$$
\vec a\cdot\vec b=2(1)+1(-2)=0
$$

Por tanto:

$$
\boxed{\vec a\perp\vec b}
$$

---

## 13. Producto vectorial

El producto vectorial habitual se define para vectores en tres dimensiones y produce **otro vector**, perpendicular a los dos originales.

$$
\boxed{\vec c=\vec a\times\vec b}
$$

### Módulo

$$
\boxed{|\vec a\times\vec b|=|\vec a||\vec b|\sin\theta}
$$

### Cálculo por componentes

Si:

$$
\vec a=(a_x,a_y,a_z)
$$

$$
\vec b=(b_x,b_y,b_z)
$$

entonces:

$$
\boxed{
\vec a\times\vec b=
\begin{vmatrix}
\vec i&\vec j&\vec k\\
a_x&a_y&a_z\\
b_x&b_y&b_z
\end{vmatrix}
}
$$

Desarrollando:

$$
\boxed{
\begin{aligned}
\vec a\times\vec b={}&(a_yb_z-a_zb_y)\vec i\\
&+(a_zb_x-a_xb_z)\vec j\\
&+(a_xb_y-a_yb_x)\vec k
\end{aligned}
}
$$

### Ejemplo

$$
\vec a=(1,0,0),\qquad\vec b=(0,1,0)
$$

$$
\boxed{\vec a\times\vec b=(0,0,1)=\vec k}
$$

El sentido del vector resultante se determina mediante la regla de la mano derecha.

### Casos importantes

- Vectores paralelos: producto vectorial nulo.
- Vectores perpendiculares: módulo del producto igual al producto de los módulos.
- Cambiar el orden invierte el resultado:

$$
\boxed{\vec a\times\vec b=-(\vec b\times\vec a)}
$$

---

## 14. Producto escalar frente a producto vectorial

| Característica | Escalar | Vectorial |
|---|---|---|
| Operación | $\vec a\cdot\vec b$ | $\vec a\times\vec b$ |
| Resultado | Número | Vector |
| Función | Coseno | Seno |
| Fórmula | $ab\cos\theta$ | Módulo $ab\sin\theta$ |
| Perpendiculares | Resultado $0$ | Módulo máximo |
| Paralelos | Valor absoluto máximo | Resultado $\vec 0$ |
| Aplicación física | Trabajo | Momento de una fuerza |

## Fórmulas imprescindibles

**Módulo en 3D**

$$
\boxed{|\vec a|=\sqrt{a_x^2+a_y^2+a_z^2}}
$$

**Vector unitario**

$$
\boxed{\hat a=\frac{\vec a}{|\vec a|}}
$$

**Componentes desde módulo y ángulo**

$$
\boxed{a_x=a\cos\alpha,\qquad a_y=a\sin\alpha}
$$

**Ángulo entre vectores**

$$
\boxed{\theta=\arccos\left(\frac{\vec a\cdot\vec b}{|\vec a||\vec b|}\right)}
$$

**Producto escalar**

$$
\boxed{\vec a\cdot\vec b=a_xb_x+a_yb_y+a_zb_z}
$$

**Producto vectorial**

$$
\boxed{|\vec a\times\vec b|=ab\sin\theta}
$$
